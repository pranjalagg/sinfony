# sinfony tour: music from mathematics

Every step below is a formula you can read, change, and hear.

```python
import sinfony as sf
from sinfony.render.play import play
from sinfony.render.synth import render
```

## 1. Curves: a shape plus noise

`arch()` is sin(πu) on [0, 1]. `pink` is 1/f noise, the kind of wandering found in real melodies. Curves add and multiply like functions.

```python
rng = sf.Rng(7)
contour = 0.6 * sf.arch() + 0.15 * sf.pink(rng, base=2)
contour.show(0, 1)
```

## 2. A scale from stacked fifths

Start on F and keep adding 7 semitones (a perfect fifth), wrapping at 12. Seven fifths give the major scale; five give the pentatonic scale.

```python
print([(7 * k) % 12 for k in range(-1, 6)])
major = sf.Scale.generated(7, 7)
penta = sf.Scale.generated(7, 5)
print(major.steps, penta.steps)
```

```python
up = sf.Events([sf.Note(i * 0.5, 0.5, major.degree(i)) for i in range(8)])
play(render(up, sf.const(120), sf.SINE))
```

## 3. Chords are sets in Z12

C major is {0, 4, 7}. The neo-Riemannian moves P, L, R each change one note by a step and keep two.

```python
C = sf.triad(major, 0)
print(C.label, sorted(C.pcs))
for move in (sf.P, sf.L, sf.R):
    print(move.__name__, move(C), sorted(move(C).pcs))
```

## 4. A melody from a contour and rules

The contour proposes a target pitch; rules (penalties) prefer steps, fill leaps, and end on the tonic.

```python
plan = sf.Plan(seed=1, form=sf.Form("A", bars=4), tempo=96)
melody = sf.Voice("melody", "melody", (60, 79), rhythm=sf.density(0.9),
                  contour=sf.arch().over(0, 16),
                  rules=[sf.step_bias(1.5), sf.gap_fill(4.0), sf.end_on_tonic(20.0)])
bass = sf.Voice("bass", "bass", (36, 55))
piece = sf.Piece(plan, sf.Fixed(["I", "IV", "V", "I"]), [melody, bass])
piece.play()
piece.show()
```

## 5. Full pieces

`tonal` builds a classical-style piece; `ambient` a slow modal one with phasing loops. Change the seed for a new piece.

```python
t = sf.tonal(seed=42)
t.play()
t.show()
```

```python
a = sf.ambient(seed=42)
a.play()
a.show()
```
